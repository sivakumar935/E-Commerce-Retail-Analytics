```python
import pandas as pd

df1 = pd.read_csv("../data/cleaned/olist_customers_dataset.csv")
df2 = pd.read_csv("../data/cleaned/olist_geolocation_dataset.csv")
df3 = pd.read_csv("../data/cleaned/olist_order_items_dataset.csv")
df4 = pd.read_csv("../data/cleaned/olist_order_payments_dataset.csv")
df5 = pd.read_csv("../data/cleaned/olist_order_reviews_dataset.csv")
df6 = pd.read_csv("../data/cleaned/olist_orders_dataset.csv")
df7 = pd.read_csv("../data/cleaned/olist_products_dataset.csv")
df8 = pd.read_csv("../data/cleaned/olist_sellers_dataset.csv")
df9 = pd.read_csv("../data/cleaned/product_category_name_translation.csv")

tables = {
    "customers": df1,
    "geolocation": df2,
    "order_items": df3,
    "order_payments": df4,
    "order_reviews": df5,
    "orders": df6,
    "products": df7,
    "sellers": df8,
    "category_translation": df9
}
```

```python
# HANDLE MISSING VALUES
for name, df in tables.items():
    print("\n", name)
    print(df.isnull().sum())
```

```python
product_numeric = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

for col in product_numeric:
    df7[col] = df7[col].fillna(df7[col].median())
```

```python
df7["product_category_name"] = df7["product_category_name"].fillna("unknown")
```

```python
df5["review_comment_title"] = df5["review_comment_title"].fillna("No comment")

df5["review_comment_message"] = df5["review_comment_message"].fillna("No comment")
```

```python
for name, df in tables.items():
    print("\n", name)
    print(df.isnull().sum())
```

```python
tables["order_reviews"] = df5
tables["products"] = df7
```

```python
# REMOVE DUPLICATES
for name, df in tables.items():
    print(name, ":", df.duplicated().sum())
```

```python
for name, df in tables.items():
    tables[name] = df.drop_duplicates()
```

```python
for name, df in tables.items():
    print(name, ":", df.duplicated().sum())
```

```python
# DATA TYPES CORRECTION
for name, df in tables.items():
    print("\n", name)
    print(df.dtypes)
```

```python
# FORMAT DATE TIME 
order_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in order_date_columns:
    df6[col] = pd.to_datetime(df6[col])
```

```python
df3["shipping_limit_date"] = pd.to_datetime(
    df3["shipping_limit_date"]
)
```

```python
review_date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

for col in review_date_columns:
    df5[col] = pd.to_datetime(df5[col])
```
